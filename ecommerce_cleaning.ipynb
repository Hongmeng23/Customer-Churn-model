```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ecommerce_churn.csv")
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (df.isnull().mean()*100).sort_values(ascending=False)

missing_report = pd.DataFrame({
    "missing_count" : missing,
    "missing_percent" : missing_pct.round(2)
})

missing_report
```

```python
for col in ["age", "annual_income", "monthly_spend"] :
    df[col] = df[col].fillna(df[col].median())

df["contract_type"] = df["contract_type"].fillna("unknown")
df.isnull().sum()
```

```python
print(df["region"].value_counts(dropna=False))
```

```python
df["region"] = (
    df["region"]
    .astype("string")
    .str.strip()
    .str.title()
)

df["region"].value_counts(dropna=False)
```

```python
print("invalid ages: ")
display(df[df["age"]<0])
```

```python
print("invalid support_call counts: ")
display(df[df["support_calls"]<0])
```

```python
df.loc[df["age"] < 0, "age"] = np.nan #makes the age that's less than 0 an empty set 
df.loc[df["support_calls"] <0, "support_calls"] = np.nan #makes the support_call that's less than 0 an empty set 

df["age"] = df["age"].fillna(df["age"].median()) #filled the empty set with a median
df["support_calls"] = df["support_calls"].fillna(df["support_calls"].median())

df[["age", "support_calls"]].describe()
```

```python
print(df["age"])
```

```python
Q1 = df["annual_income"].quantile(0.25)
Q3 = df["annual_income"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 - 1.5 * IQR

outliers = df[
    (df["annual_income"] < lower) |
    (df["annual_income"] > upper)
]

print("Lower bound:", lower)
print("Upper bound:", upper)

display(outliers)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values("customer_id")
```

```python
df = df.drop_duplicates()
print("Number of duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print("Missing Values:")
display(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())
print("Negative ages:", (df["age"] < 0).sum())
print("Negayive support calls:", (df["support_calls"] < 0).sum())

print("\nRegions:")
display(df["region"].value_counts())
```

```python
df.to_csv("ecommerce_churn_cleaned.csv", index=False)
print("saved: ", "ecommerce_churn_cleaned.csv")
print("Final shape: ", df.shape)
```
